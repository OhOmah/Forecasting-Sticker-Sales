```python
# Initial plot done, just need to clean it up
plt.figure(figsize=(15,4))
sns.set_style("darkgrid")
ax = sns.lineplot(
    data=train, 
    x="date", 
    y="num_sold", 
    hue='country',
    errorbar=None,
    linewidth=0.3,
    legend=True)
ax.set_xlabel("Year", fontsize=10)
ax.tick_params(axis='both', labelsize=8)
plt.show()
```
